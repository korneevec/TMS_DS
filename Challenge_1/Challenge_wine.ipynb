# Добавление шума

```python
def add_noise(x):
    noise = np.random.uniform(-0.8, 0.8)
    if x == 1:
        return x + abs(noise)
    elif x == 10:
        return x - abs(noise)
    else:
        return x + noise

wine_df['quality'] = wine_df['quality'].apply(add_noise)
```
