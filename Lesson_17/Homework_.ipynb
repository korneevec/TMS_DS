```python
# Импортируем необходимые библиотеки
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt

# Загрузим данные
fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Нормализуем данные
train_images = train_images / 255.0
test_images = test_images / 255.0

# Создаем модель
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10)
])

# Компилируем модель
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# Обучаем модель
model.fit(train_images, train_labels, epochs=10)

# Оценим точность модели
test_loss, test_acc = model.evaluate(test_images,  test_labels, verbose=2)
print('\nТестовая точность:', test_acc)
```

8) Используйте разные виды инициализации весов (например, he_normal или glorot_uniform) и сравните результаты.

```python
# Импортируем необходимые библиотеки
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
from tensorflow.keras.initializers import HeNormal, GlorotUniform

# Загрузим данные
fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Нормализуем данные
train_images = train_images / 255.0
test_images = test_images / 255.0

# Создаем модель
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu', kernel_initializer=HeNormal()), # HeNormal инициализатор
    keras.layers.Dense(10, kernel_initializer=GlorotUniform()) 
])

# Компилируем модель
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# Обучаем модель
model.fit(train_images, train_labels, epochs=10)

# Оценим точность модели
test_loss, test_acc = model.evaluate(test_images,  test_labels, verbose=2)
print('\nТестовая точность:', test_acc)
```

9) Используйте регуляризацию L1 или L2 и сравните результаты.

```python
# Импортируем необходимые библиотеки
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
from tensorflow.keras import regularizers

# Загрузим данные
fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Нормализуем данные
train_images = train_images / 255.0
test_images = test_images / 255.0

# Создаем модель
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu', 
                      kernel_regularizer=regularizers.l2(0.01)),
    keras.layers.Dense(10)
])

# Компилируем модель
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# Обучаем модель
model.fit(train_images, train_labels, epochs=10)

# Оценим точность модели
test_loss, test_acc = model.evaluate(test_images,  test_labels, verbose=2)
print('\nТестовая точность:', test_acc)
```
